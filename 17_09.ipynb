```python
import matplotlib.pyplot as plt
df = pd.DataFrame({
      "ay": ["Ocak", "Şubat","Mart","Nisan"],
      "satis":[1200,1800,900,1500]
})
```

```python
#çizgi grafik
df.plot(x="ay",y="satis",kind="line", marker="o")
plt.title("Ay Bazlı Satış")
plt.xlabel("Ay")
plt.ylabel("Satış Miktarı")
plt.show()
```

```python
df.plot(x="ay",y="satis",kind="bar", color ="skyblue")
plt.title("Ay Bazlı Satış")
plt.xlabel("Ay")
plt.ylabel("Satış Miktarı")
plt.show()
```

```python
df.set_index("ay").plot(kind="pie",y="satis",autopct="%1.1f%%")
plt.title("Ay Bazlı Satış")
plt.show()
```

```python
df.plot(x="ay",y="satis",kind="scatter", marker="o")
plt.title("Ay Bazlı Satış")
plt.xlabel("Ay")
plt.ylabel("Satış Miktarı")
plt.show()
```
