---## Bölüm 6 — Filtreleme ve Yeni Sütun**Soru 6.1:** Toplam tutarı 300 TL'nin üzerinde olan siparişleri filtreleyin.**Soru 6.2:** Ürün adedi 2'den fazla VE toplam tutar 200 TL'den fazla olan siparişleri bulun (`&` kullanın).**Soru 6.3:** `birim_fiyat` adında yeni bir sütun ekleyin; değeri `toplam_tutar / urun_adedi` olsun.**Soru 6.4:** `apply` ve `lambda` kullanarak kargo süresi 3 günden fazlaysa "Yavaş", değilse "Hızlı" yazan bir `kargo_durumu` sütunu oluşturun.

```python
# 6.1
print(df[df.toplam_tutar > 300])

# 6.2
print(df[(df.urun_adedi > 2) & (df.toplam_tutar > 200)])

# 6.3
df["birim_fiyat"] = df.toplam_tutar / df.urun_adedi
print(df)

# 6.4
df["kargo_durumu"] = df.kargo_suresi.apply(lambda x: "Yavaş" if x > 3 else "Hızlı")
print(df)
```

---## Bölüm 9 — Merge (Tabloları Birleştirme)Aşağıdaki iki tabloyu kullanın:

```python
musteriler = pd.DataFrame({
    "musteri_id": [1, 2, 3, 4],
    "ad": ["Baran", "Ceylin", "Emre", "Fulya"]
})

siparisler = pd.DataFrame({
    "musteri_id": [1, 2, 5],
    "urun": ["Kulaklık", "Şarj Kablosu", "Powerbank"],
    "tutar": [450, 90, 320]
})
```

**Soru 9.1:** `musteriler` ve `siparisler` tablolarını **inner join** ile birleştirin — kaç müşterinin siparişi eşleşiyor?**Soru 9.2:** Aynı tabloları **left join** ile birleştirin — hiç siparişi olmayan müşteri(ler) sonuçta ne görünüyor?**Soru 9.3:** Aynı tabloları **outer join** ile birleştirip, hem eşleşmeyen müşterileri hem eşleşmeyen siparişleri gözlemleyin.

```python
# 9.1 inner join (ortak kayitlar)
print("Inner Join:\n", pd.merge(musteriler, siparisler, on="musteri_id", how="inner"))

# 9.2 left join (musterilerin tamami, eslesmeyenler NaN)
print("\nLeft Join:\n", pd.merge(musteriler, siparisler, on="musteri_id", how="left"))

# 9.3 outer join (butun musteriler ve butun siparisler)
print("\nOuter Join:\n", pd.merge(musteriler, siparisler, on="musteri_id", how="outer"))
```

---## Bölüm 10 — Concat**Soru 10.1:** `hafta1 = pd.DataFrame({"gun":["Pzt","Sal"], "adet":[20,35]})` ve `hafta2 = pd.DataFrame({"gun":["Çar","Per"], "adet":[28,40]})` tablolarını `ignore_index=True` ile alt alta birleştirin.**Soru 10.2:** `gelir = pd.DataFrame({"toplam_gelir":[1000,1750]})` tablosunu, `hafta1` tablosuyla `axis=1` kullanarak yan yana birleştirin.

```python
hafta1 = pd.DataFrame({"gun": ["Pzt", "Sal"], "adet": [20, 35]})
hafta2 = pd.DataFrame({"gun": ["Çar", "Per"], "adet": [28, 40]})
gelir = pd.DataFrame({"toplam_gelir": [1000, 1750]})

# 10.1 alt alta birlestirme
print("Alt alta:\n", pd.concat([hafta1, hafta2], ignore_index=True))

# 10.2 yan yana sutun ekleme
print("\nYan yana:\n", pd.concat([hafta1, gelir], axis=1))
```

---## Bölüm 11 — GroupByAşağıdaki veriyi kullanın:

```python
df_calisan = pd.DataFrame({
    "sube": ["Kadıköy", "Beşiktaş", "Kadıköy", "Şişli", "Beşiktaş"],
    "ad": ["Baran", "Ceylin", "Emre", "Fulya", "Gökhan"],
    "satis": [12000, 18500, 9500, 21000, 16000]
})
```

**Soru 11.1:** Şubeye göre **toplam** satışı bulun.**Soru 11.2:** Şubeye göre **ortalama** satışı bulun.**Soru 11.3:** `.agg(["sum","mean","count"])` kullanarak şube başına üç istatistiği aynı anda hesaplayın.

```python
# 11.1 sube toplam satis
print("Şube toplam:\n", df_calisan.groupby("sube")["satis"].sum())

# 11.2 sube ortalama satis
print("\nŞube ortalama:\n", df_calisan.groupby("sube")["satis"].mean())

# 11.3 agg ile coklu istatistik
print("\nÖzet:\n", df_calisan.groupby("sube")["satis"].agg(["sum", "mean", "count"]))
```

---## Bölüm 12 — Pivot TableAşağıdaki veriyi kullanın:

```python
df_satis = pd.DataFrame({
    "sehir": ["Bursa", "Adana", "Antalya", "Bursa", "Adana"],
    "kategori": ["Elektronik", "Giyim", "Elektronik", "Giyim", "Elektronik"],
    "satis": [5000, 3200, 4100, 2800, 3900]
})
```

**Soru 12.1:** `index="sehir"`, `columns="kategori"`, `values="satis"` ile bir pivot tablo oluşturun.**Soru 12.2:** Aynı pivot tabloyu `aggfunc="sum"` ve `fill_value=0` parametreleriyle oluşturup eksik hücrelerin nasıl doldurulduğunu gözlemleyin.

```python
# 12.1 temel pivot tablo
print(df_satis.pivot_table(index="sehir", columns="kategori", values="satis"))

# 12.2 sum ve fill_value=0 ile
print("\n", df_satis.pivot_table(index="sehir", columns="kategori", values="satis", aggfunc="sum", fill_value=0))
```

---## Bölüm 13 — Apply ile String İşlemleri**Soru 13.1:** `df_kullanici = pd.DataFrame({"kullanici_adi": ["baran23", "CEYLIN_K", "emre.y"]})` oluşturun.**Soru 13.2:** `apply(str.lower)` kullanarak tüm kullanıcı adlarını küçük harfe çeviren bir `kullanici_adi_kucuk` sütunu ekleyin.**Soru 13.3:** `apply` ve `lambda` ile kullanıcı adının kaç karakter olduğunu gösteren bir `karakter_sayisi` sütunu ekleyin.

```python
# 13.1 df olusturma
df_kullanici = pd.DataFrame({"kullanici_adi": ["baran23", "CEYLIN_K", "emre.y"]})

# 13.2 kucuk harfe cevirme
df_kullanici["kullanici_adi_kucuk"] = df_kullanici["kullanici_adi"].apply(str.lower)

# 13.3 karakter sayisi bulma
df_kullanici["karakter_sayisi"] = df_kullanici["kullanici_adi"].apply(lambda x: len(x))
print(df_kullanici)
```

---## Bölüm 15 — Matplotlib ile Grafik ÇizmeAşağıdaki hücreyi çalıştırarak bir yıllık çeyreklik gelir verisi oluşturun:

```python
df_gelir = pd.DataFrame({
    "ceyrek": ["Ç1", "Ç2", "Ç3", "Ç4"],
    "gelir": [42000, 55000, 48000, 61000],
    "gider": [30000, 38000, 33000, 40000]
})
```

**Soru 15.1:** `gelir` değerlerini `kind="line"` ile, üzerinde nokta işaretleri (`marker="o"`) olacak şekilde çizgi grafikle çizin; başlık ve eksen etiketleri ekleyin.**Soru 15.2:** `gelir` değerlerini çeyreklere göre `kind="bar"` ile bar grafikle çizin (renk seçiminiz size ait).**Soru 15.3:** `gelir` sütununu çeyreklerin payını gösterecek şekilde `kind="pie"` ile pasta grafikle çizin (`autopct="%1.1f%%"` kullanın).**Soru 15.4:** `gelir` ve `gider` sütunlarını `kind="scatter"` ile scatter grafikte karşılaştırın (x ekseni gelir, y ekseni gider olsun) ve grafiğin ne anlattığını bir cümleyle yorumlayın.

```python
# 15.1 cizgi grafigi
df_gelir.plot(kind="line", x="ceyrek", y="gelir", marker="o", title="Çeyreklik Gelir Grafiği")
plt.xlabel("Çeyrekler")
plt.ylabel("Gelir (TL)")
plt.show()

# 15.2 bar grafigi
df_gelir.plot(kind="bar", x="ceyrek", y="gelir", title="Yıllık Gelir", color="teal")
plt.xlabel("Çeyrekler")
plt.ylabel("Gelir (TL)")
plt.show()

# 15.3 pasta grafigi
df_gelir.plot(kind="pie", y="gelir", labels=df_gelir["ceyrek"], autopct="%1.1f%%", title="Yıllık Gelir Dağılımı")
plt.ylabel("")
plt.show()

# 15.4 scatter grafigi
df_gelir.plot(kind="scatter", x="gelir", y="gider", title="Yıllık Gelir ve Gider İlişkisi")
plt.xlabel("Gelir (TL)")
plt.ylabel("Gider (TL)")
plt.show()
# Yorum: Gelir arttıkça giderin de paralel olarak arttığı doğrusal ve pozitif bir ilişki görülmektedir.
```

---## Bonus — Kapsamlı UygulamaBir online kursun satış verisini simüle edin:1. `kurs_adi`, `kategori` (örn. Yazılım, Tasarım, Pazarlama) ve `satis_adedi` sütunlarından oluşan, en az 8 satırlık bir DataFrame oluşturun.2. Kategoriye göre toplam satış adedini `groupby` ile bulun.3. En çok satan 3 kursu `.sort_values()` ile bulun.4. Kategori bazında toplam satışı gösteren bir bar grafik çizin.5. `apply` ile satış adedi 50 üzeri olanlara "Popüler", altındakilere "Standart" yazan yeni bir sütun ekleyin.

```python
# 1. kurs tablosu simulasyonu
df_kurs = pd.DataFrame({
    "kurs_adi": [
        "Sıfırdan İleri Seviye Python",
        "SQL ile Veri Tabanı Yönetimi",
        "Figma ile Mobil Arayüz Tasarımı",
        "Photoshop Temelleri",
        "Arama Motoru Optimizasyonu (SEO)",
        "Sosyal Medya Reklamcılığı",
        "React ile Web Geliştirme",
        "İş Zekası ve Raporlama"
    ],
    "kategori": [
        "Yazılım", "Yazılım", "Tasarım", "Tasarım",
        "Pazarlama", "Pazarlama", "Yazılım", "Yazılım"
    ],
    "satis_adedi": [120, 85, 45, 30, 95, 40, 110, 65]
})
print(df_kurs)

# 2. kategori bazinda toplam satis
kategori_ozet = df_kurs.groupby("kategori")["satis_adedi"].sum()
print("\nKategori toplam satışları:\n", kategori_ozet)

# 3. en cok satan 3 kurs
print("\nEn çok satan 3 kurs:\n", df_kurs.sort_values("satis_adedi", ascending=False).head(3))

# 4. kategori bazinda bar grafik
kategori_ozet.plot(kind="bar", title="Kategori Bazında Toplam Satış", color="royalblue")
plt.xlabel("Kategori")
plt.ylabel("Toplam Satış Adedi")
plt.show()

# 5. populer / standart etiketi
df_kurs["durum"] = df_kurs["satis_adedi"].apply(lambda x: "Popüler" if x > 50 else "Standart")
print("\nSon durum tablosu:\n", df_kurs)
```
